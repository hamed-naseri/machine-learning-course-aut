# Classification Metrics

چهار Metric اصلی:

<bdi dir="ltr">Accuracy</bdi>،
<bdi dir="ltr">Precision</bdi>،
<bdi dir="ltr">Recall</bdi>،
<bdi dir="ltr">F1-score</bdi>

</div>



<div dir="rtl">

## ۱. Confusion Matrix


<div style="text-align:center; margin:6px 0;">

</div>


- <bdi dir="ltr">TP</bdi>: Positive واقعی و Prediction هم Positive
- <bdi dir="ltr">TN</bdi>: Negative واقعی و Prediction هم Negative
- <bdi dir="ltr">FP</bdi>: Positive اشتباه
- <bdi dir="ltr">FN</bdi>: Positive واقعی که از دست رفته

</div>



<div dir="rtl">

## ۲. Formulaها

<div class="formula"><b>Accuracy = (TP + TN) / (TP + TN + FP + FN)</b></div>

<div class="formula"><b>Precision = TP / (TP + FP)</b></div>

<div class="formula"><b>Recall = TP / (TP + FN)</b></div>

<div class="formula"><b>F1 = 2 × Precision × Recall / (Precision + Recall)</b></div>


<div style="text-align:center; margin:6px 0;">

</div>


</div>



<div dir="rtl">

## ۳. مثال فرضی COVID

**مدل A — سخت‌گیر**

<bdi dir="ltr">TP=20, FN=80, FP=2, TN=898</bdi>

- Precision ≈ **91%**
- Recall = **20%**

**مدل B — حساس‌تر**

<bdi dir="ltr">TP=90, FN=10, FP=100, TN=800</bdi>

- Precision ≈ **47%**
- Recall = **90%**


<div style="text-align:center; margin:6px 0;">

</div>


</div>



<div dir="rtl">

## ۴. خلاصه

| Metric | سؤال اصلی |
|---|---|
| Accuracy | چند درصد کل Predictionها درست بود؟ |
| Precision | وقتی Positive گفت، چند درصد درست بود؟ |
| Recall | از Positiveهای واقعی چند درصد پیدا شدند؟ |
| F1 | Precision و Recall با هم چقدر خوب‌اند؟ |

**Precision → Predicted Positive**

**Recall → Actual Positive**

</div>
